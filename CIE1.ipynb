## 3. Data Transformation

In this section, we perform data transformation by removing unnecessary columns, handling duplicate records, changing data types and creating new columns.

```python
print(df[['Unnamed: 0']].head(10))
```

```python
df = df.drop(columns=['Unnamed: 0'])

print("Unnecessary column removed successfully!")
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print("Duplicate rows removed successfully!")
```

```python
df['duration_min'] = df['duration_ms'] / 60000
print(df[['track_name', 'duration_ms', 'duration_min']].head())
```

```python
print("Updated Dataset Shape:", df.shape)
print("Updated Column Names:")
print(df.columns.tolist())
```

## 4. Handling Null Values

In this section, we identify missing values in the
Spotify dataset and handle numerical missing values
using mean imputation.

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

print("Numerical Columns:")
print(numeric_cols.tolist())

print("Missing Values in Numerical Columns:")
print(df[numeric_cols].isnull().sum())
```

```python
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())
print("Missing numerical values filled using mean imputation!")
```

```python
print("Missing Values After Imputation:")
print(df.isnull().sum())
```

## 5. Insight Generation

In this section, we analyse the Spotify dataset to identify meaningful patterns and trends related to song popularity, artists, genres and music characteristics.

```python
top_songs = df.nlargest(10, 'popularity')
print("Top 10 Most Popular Songs:")
print(top_songs[['track_name', 'artists', 'popularity']])
```

```python
genre_popularity = df.groupby('track_genre')['popularity'].mean()
genre_popularity = genre_popularity.sort_values(ascending=False)
print("Top 10 Popular Genres:")
print(genre_popularity.head(10))
```

```python
artist_popularity = df.groupby('artists')['popularity'].mean()
artist_popularity = artist_popularity.sort_values(ascending=False)
print("Top 10 Artists by Average Popularity:")
print(artist_popularity.head(10))
```

```python
average_duration = df['duration_min'].mean()
print("Average Song Duration:", round(average_duration, 2), "minutes")
```

```python
print("Average Danceability:", round(df['danceability'].mean(), 2))
print("Average Energy:", round(df['energy'].mean(), 2))
```

## 6. Data Visualization

In this section, we use Matplotlib and Seaborn to visualize the insights generated from the Spotify dataset.

```python
plt.figure(figsize=(12, 6))

top_songs = df.nlargest(10, 'popularity')

sns.barplot(data=top_songs, x='track_name', y='popularity')

plt.title('Top 10 Most Popular Songs on Spotify')
plt.xlabel('Song Name')
plt.ylabel('Popularity')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

top_genres = genre_popularity.head(10)

sns.barplot(x=top_genres.values, y=top_genres.index)

plt.title('Top 10 Most Popular Music Genres')
plt.xlabel('Average Popularity')
plt.ylabel('Music Genre')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df['duration_min'], bins=30)

plt.title('Distribution of Song Duration')
plt.xlabel('Song Duration (Minutes)')
plt.ylabel('Number of Songs')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(data=df, x='danceability', y='energy', alpha=0.4)

plt.title('Relationship Between Danceability and Energy')
plt.xlabel('Danceability')
plt.ylabel('Energy')
plt.tight_layout()
plt.show()
```
